# Ranking Stage: XGBoost

This notebook:
1. Loads ratings and feature tables
2. Generates ranking training set (positive + negative samples)
3. Joins with feature store
4. Trains XGBoost ranking model
5. Saves preprocessing pipeline + model


```python
import pandas as pd
import numpy as np
import joblib
from pathlib import Path
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import xgboost as xgb

# Set up paths
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data"
PROCESSED_DIR = DATA_DIR / "processed"
MODELS_DIR = PROJECT_ROOT / "models"
MODELS_DIR.mkdir(exist_ok=True)

print(f"Project root: {PROJECT_ROOT}")
print(f"Processed directory: {PROCESSED_DIR}")
print(f"Models directory: {MODELS_DIR}")
```

```python
# Load data
ratings_path = PROCESSED_DIR / "ratings_clean.parquet"
user_features_path = PROCESSED_DIR / "user_features.parquet"
item_features_path = PROCESSED_DIR / "item_features.parquet"

print("Loading data...")
df_ratings = pd.read_parquet(ratings_path)
df_user_features = pd.read_parquet(user_features_path)
df_item_features = pd.read_parquet(item_features_path)

print(f"Ratings shape: {df_ratings.shape}")
print(f"User features shape: {df_user_features.shape}")
print(f"Item features shape: {df_item_features.shape}")

print("\nUser features columns:", df_user_features.columns.tolist())
print("Item features columns:", df_item_features.columns.tolist())
```

```python
# Generate ranking training set
print("Generating ranking training set...")

# Positive samples: actual interactions
positives = df_ratings[['user_id', 'parent_asin', 'rating']].copy()
positives['target'] = positives['rating'].astype(float)
positives['label'] = 1

print(f"Positive samples: {len(positives)}")

# Negative samples: random items user has NOT interacted with
neg_per_pos = 3
print(f"Generating {neg_per_pos} negative samples per positive...")

# Set random seed for reproducibility
np.random.seed(42)

# Get all unique items
all_items = set(df_item_features['parent_asin'].unique())

# Get user->seen items mapping
user_seen_items = {}
for _, row in df_ratings.iterrows():
    user_id = row['user_id']
    item_asin = row['parent_asin']
    if user_id not in user_seen_items:
        user_seen_items[user_id] = set()
    user_seen_items[user_id].add(item_asin)

# Generate negatives
negatives_list = []
for user_id, seen_items in user_seen_items.items():
    unseen_items = list(all_items - seen_items)
    n_negatives = min(neg_per_pos * len(seen_items), len(unseen_items))
    if n_negatives > 0:
        sampled_unseen = np.random.choice(unseen_items, size=n_negatives, replace=False)
        for item_asin in sampled_unseen:
            negatives_list.append({
                'user_id': user_id,
                'parent_asin': item_asin,
                'target': 0.0,
                'label': 0
            })

negatives = pd.DataFrame(negatives_list)
print(f"Negative samples: {len(negatives)}")

# Combine
df_ranking = pd.concat([positives[['user_id', 'parent_asin', 'target', 'label']], 
                        negatives], ignore_index=True)

print(f"Total ranking samples: {len(df_ranking)}")
print(f"Positive ratio: {df_ranking['label'].mean():.3f}")
```

```python
# Join with feature store
print("Joining with feature store...")

# Join user features
df_ranking = df_ranking.merge(df_user_features, on='user_id', how='left')

# Join item features
df_ranking = df_ranking.merge(df_item_features, on='parent_asin', how='left')

print(f"After joins - shape: {df_ranking.shape}")
print(f"Columns: {df_ranking.columns.tolist()}")
print(f"\nMissing values:")
print(df_ranking.isnull().sum())
```

```python
# Prepare features
print("Preparing features...")

# Separate target
y = df_ranking['target'].values

# Feature columns
numeric_features = ['user_avg_rating', 'total_user_reviews', 'average_item_rating', 'price_float']
categorical_features = ['brand', 'main_category']

# Check which features exist
available_numeric = [f for f in numeric_features if f in df_ranking.columns]
available_categorical = [f for f in categorical_features if f in df_ranking.columns]

print(f"Available numeric features: {available_numeric}")
print(f"Available categorical features: {available_categorical}")

# Fill NaN for numeric with median
for col in available_numeric:
    median_val = df_ranking[col].median()
    df_ranking[col] = df_ranking[col].fillna(median_val)

# Fill NaN for categorical with "unknown"
for col in available_categorical:
    df_ranking[col] = df_ranking[col].fillna("unknown")

# Prepare feature matrix
X = df_ranking[available_numeric + available_categorical].copy()

print(f"Feature matrix shape: {X.shape}")
print(f"Target shape: {y.shape}")
```

```python
# Create preprocessing pipeline
print("Creating preprocessing pipeline...")

# Numeric imputer (median)
numeric_imputer = SimpleImputer(strategy='median')

# Categorical imputer (constant "unknown")
categorical_imputer = SimpleImputer(strategy='constant', fill_value='unknown')

# One-hot encoder for categorical
categorical_encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Column transformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_imputer, available_numeric),
        ('cat', categorical_encoder, available_categorical)
    ],
    remainder='drop'
)

# Fit and transform
print("Fitting preprocessor...")
X_processed = preprocessor.fit_transform(X)

print(f"Processed feature matrix shape: {X_processed.shape}")
print(f"Number of features after encoding: {X_processed.shape[1]}")
```

```python
# Split train/validation
print("Splitting data...")
X_train, X_val, y_train, y_val = train_test_split(
    X_processed, y, test_size=0.2, random_state=42, stratify=None
)

print(f"Train size: {X_train.shape[0]}")
print(f"Validation size: {X_val.shape[0]}")
```

```python
# Train XGBoost Regressor
print("Training XGBoost model...")

model = xgb.XGBRegressor(
    n_estimators=400,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.9,
    colsample_bytree=0.9,
    random_state=42,
    n_jobs=-1
)

model.fit(
    X_train, y_train,
    eval_set=[(X_val, y_val)],
    verbose=50
)

print("Training complete!")
```

```python
# Evaluate
y_train_pred = model.predict(X_train)
y_val_pred = model.predict(X_val)

train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
val_rmse = np.sqrt(mean_squared_error(y_val, y_val_pred))

print(f"Train RMSE: {train_rmse:.4f}")
print(f"Validation RMSE: {val_rmse:.4f}")
```

```python
# Save pipeline (preprocessor + model)
pipeline_path = MODELS_DIR / "xgb_ranker.joblib"

pipeline = {
    'preprocessor': preprocessor,
    'model': model,
    'numeric_features': available_numeric,
    'categorical_features': available_categorical
}

joblib.dump(pipeline, pipeline_path)

print(f"Saved pipeline to: {pipeline_path}")
print(f"Pipeline file size: {pipeline_path.stat().st_size / (1024*1024):.2f} MB")
print("\nRanking model training complete!")
```
