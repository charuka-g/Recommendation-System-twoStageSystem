# Retrieval Stage: Surprise SVD

This notebook:
1. Loads cleaned ratings data
2. Trains a Surprise SVD model
3. Implements candidate retrieval function
4. Saves the trained model


```python
import pandas as pd
import pickle
from pathlib import Path
from surprise import SVD, Dataset, Reader
from collections import defaultdict

# Set up paths
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data"
PROCESSED_DIR = DATA_DIR / "processed"
MODELS_DIR = PROJECT_ROOT / "models"
MODELS_DIR.mkdir(exist_ok=True)

print(f"Project root: {PROJECT_ROOT}")
print(f"Processed directory: {PROCESSED_DIR}")
print(f"Models directory: {MODELS_DIR}")
```

```python
# Load cleaned ratings
ratings_path = PROCESSED_DIR / "ratings_clean.parquet"
df_ratings = pd.read_parquet(ratings_path)

print(f"Ratings shape: {df_ratings.shape}")
print(f"Columns: {df_ratings.columns.tolist()}")
print(f"Unique users: {df_ratings['user_id'].nunique()}")
print(f"Unique items: {df_ratings['parent_asin'].nunique()}")
print(f"\nFirst few rows:")
print(df_ratings.head())
```

```python
# Prepare data for Surprise
reader = Reader(rating_scale=(1, 5))
data = Dataset.load_from_df(df_ratings[['user_id', 'parent_asin', 'rating']], reader)

# Build full trainset
trainset = data.build_full_trainset()

print(f"Number of users: {trainset.n_users}")
print(f"Number of items: {trainset.n_items}")
print(f"Number of ratings: {trainset.n_ratings}")
```

```python
# Train SVD model
print("Training SVD model...")
model = SVD(n_factors=50, n_epochs=20, lr_all=0.005, reg_all=0.02, random_state=42)
model.fit(trainset)

print("Model training complete!")
```

```python
# Precompute user->seen items mapping and all items list for efficiency
print("Precomputing user-item mappings...")

user_seen_items = defaultdict(set)
for _, row in df_ratings.iterrows():
    user_seen_items[row['user_id']].add(row['parent_asin'])

all_items = set(df_ratings['parent_asin'].unique())

print(f"Total unique items: {len(all_items)}")
print(f"Users with interactions: {len(user_seen_items)}")
```

```python
def get_candidates(user_id: str, model, trainset, user_seen_items: dict, all_items: set, n: int = 100):
    """
    Get top-n candidate items for a user using SVD predictions.
    
    Args:
        user_id: User identifier
        model: Trained Surprise SVD model
        trainset: Surprise trainset
        user_seen_items: Dict mapping user_id to set of seen item ASINs
        all_items: Set of all item ASINs
        n: Number of candidates to return
    
    Returns:
        List of tuples [(parent_asin, predicted_score)] sorted by score descending
    """
    # Get items user has NOT interacted with
    seen = user_seen_items.get(user_id, set())
    unseen_items = all_items - seen
    
    if len(unseen_items) == 0:
        return []
    
    # Get user internal ID
    try:
        user_inner_id = trainset.to_inner_uid(user_id)
    except ValueError:
        # User not in training set (cold start)
        return []
    
    # Predict ratings for all unseen items
    predictions = []
    for item_asin in unseen_items:
        try:
            item_inner_id = trainset.to_inner_iid(item_asin)
            pred = model.predict(user_inner_id, item_inner_id)
            predictions.append((item_asin, pred.est))
        except ValueError:
            # Item not in training set, skip
            continue
    
    # Sort by predicted score descending and return top-n
    predictions.sort(key=lambda x: x[1], reverse=True)
    return predictions[:n]

# Test the function
test_user = df_ratings['user_id'].iloc[0]
print(f"Testing with user: {test_user}")
candidates = get_candidates(test_user, model, trainset, user_seen_items, all_items, n=10)
print(f"Retrieved {len(candidates)} candidates")
print("Top 5 candidates:")
for asin, score in candidates[:5]:
    print(f"  {asin}: {score:.4f}")
```

```python
# Save model and helper data
model_path = MODELS_DIR / "svd_surprise.pkl"

# Create a dictionary with model and helper data
model_data = {
    'model': model,
    'trainset': trainset,
    'user_seen_items': dict(user_seen_items),  # Convert defaultdict to dict
    'all_items': list(all_items)  # Convert set to list for pickle
}

with open(model_path, 'wb') as f:
    pickle.dump(model_data, f)

print(f"Saved model and helper data to: {model_path}")
print(f"Model file size: {model_path.stat().st_size / (1024*1024):.2f} MB")
```
